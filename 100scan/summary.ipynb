```python
import os

def list_yaml_files(folder_path):
    """
    递归列出指定文件夹及其子文件夹下所有以 .yaml 结尾的文件。

    Args:
        folder_path (str): 文件夹路径。

    Returns:
        list: 所有以 .yaml 结尾的文件的完整路径列表。
    """
    sum=0
    yaml_files = []
    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith(".yaml"):  # 筛选出 .yaml 文件
                yaml_files.append(os.path.join(root, file))
                sum+=1

    print(f"总共找到 {sum} 个 .mql 文件。")
    return yaml_files

# 指定文件夹路径
folder_path = "/Users/djoker/Desktop/CTGA/demo/input/semgrep/python_classified_rules"
# 调用函数并打印结果
yaml_file_names = list_yaml_files(folder_path)
print(f"文件夹 '{folder_path}' 及其子文件夹下的所有 .yaml 文件如下：")
for file_name in yaml_file_names:
    print(file_name)
```

```python
import os
import csv

def compare_folders_with_csv(folder_path, csv_path):
    """
    对比文件夹下的文件夹名与 CSV 文件中的 name 列，找出缺失的项。

    Args:
        folder_path (str): 文件夹路径。
        csv_path (str): CSV 文件路径。

    Returns:
        list: CSV 中存在但文件夹中缺失的 name 列项。
    """
    # 获取文件夹下的所有文件夹名
    folder_names = {name for name in os.listdir(folder_path) if os.path.isdir(os.path.join(folder_path, name))}
    
    # 读取 CSV 文件中的 name 列
    csv_names = set()
    with open(csv_path, mode='r', encoding='utf-8') as csv_file:
        reader = csv.DictReader(csv_file)
        for row in reader:
            csv_names.add(row['name'])
    
    # 找出 CSV 中存在但文件夹中缺失的项
    missing_names = csv_names - folder_names
    return missing_names

# 指定文件夹路径和 CSV 文件路径
folder_path = "/Users/djoker/Desktop/CTGA/demo/100scan/data/repo"  # 替换为你的文件夹路径
csv_path = "/Users/djoker/Desktop/CTGA/demo/100scan/data/100.csv"   # 替换为你的 CSV 文件路径

# 调用函数并打印结果
missing = compare_folders_with_csv(folder_path, csv_path)
if missing:
    print("以下 name 在文件夹中缺失：")
    for name in missing:
        print(name)
else:
    print("所有 CSV 中的 name 都有对应的文件夹。")
```

```python
import os
import csv

def remove_prefix_and_filter_csv(folder_path, output_folder_path, prefix_to_remove):
    """
    遍历文件夹中的所有 .csv 文件，删除 filename 列中的指定前缀，
    并筛选掉 filename 列中包含 'test' 的行，将结果保存到另一个文件夹中。
    
    Args:
        folder_path (str): 源文件夹路径。
        output_folder_path (str): 目标文件夹路径。
        prefix_to_remove (str): 要删除的前缀。
    """
    # 确保目标文件夹存在
    os.makedirs(output_folder_path, exist_ok=True)
    
    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            output_file_path = os.path.join(output_folder_path, filename)
            print(f"正在处理文件: {file_path}")
            
            # 读取并修改文件内容
            with open(file_path, mode="r", encoding="utf-8") as infile:
                reader = csv.DictReader(infile)
                rows = []
                for row in reader:
                    if "filename" in row:
                        # 如果 filename 列中包含 'test'，跳过该行
                        if "test" in row["filename"]:
                            continue
                        # 删除指定前缀
                        row["filename"] = row["filename"].removeprefix(prefix_to_remove)
                    rows.append(row)
                
            # 将修改后的内容写入目标文件夹
            with open(output_file_path, mode="w", encoding="utf-8", newline="") as outfile:
                writer = csv.DictWriter(outfile, fieldnames=reader.fieldnames)
                writer.writeheader()
                writer.writerows(rows)
            
            print(f"文件已处理并保存到: {output_file_path}")

# 指定源文件夹路径和目标文件夹路径
folder_path = "/Users/djoker/Desktop/CTGA/demo/100scan/data/bandit_results"  # 替换为你的源文件夹路径
output_folder_path = "/Users/djoker/Desktop/CTGA/demo/100scan/data/modified_results"  # 替换为你的目标文件夹路径
prefix_to_remove = "/Users/djoker/Desktop/CTGA/demo/100scan/data/repo/"

# 调用函数处理 CSV 文件
remove_prefix_and_filter_csv(folder_path, output_folder_path, prefix_to_remove)
```

```python
import os
import csv

def count_csv_rows_in_folder(folder_path):
    """
    统计文件夹中每个 CSV 文件的行数（不包括表头）。

    Args:
        folder_path (str): 文件夹路径。

    Returns:
        dict: 每个 CSV 文件的行数，以及总行数。
    """
    csv_row_counts = {}
    total_rows = 0

    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            print(f"正在统计文件: {file_path}")
            
            # 统计行数
            with open(file_path, mode="r", encoding="utf-8") as csv_file:
                reader = csv.reader(csv_file)
                row_count = sum(1 for _ in reader) - 1  # 减去表头
                csv_row_counts[filename] = row_count
                total_rows += row_count

    return csv_row_counts, total_rows

def write_sorted_counts_to_file(csv_row_counts, output_file):
    """
    将统计结果按降序排序并写入文件。

    Args:
        csv_row_counts (dict): 每个 CSV 文件的行数。
        output_file (str): 输出文件路径。
    """
    # 按行数降序排序
    sorted_counts = sorted(csv_row_counts.items(), key=lambda x: x[1], reverse=True)
    
    with open(output_file, mode="w", encoding="utf-8") as file:
        for filename, row_count in sorted_counts:
            file.write(f"{filename}: {row_count} 条记录\n")
        file.write(f"\n总记录数: {sum(csv_row_counts.values())} 条\n")

# 指定文件夹路径和输出文件路径
folder_path = "/Users/djoker/Desktop/CTGA/demo/100scan/data/modified_results"  # 替换为你的文件夹路径
output_file = "/Users/djoker/Desktop/CTGA/demo/100scan/data/output.txt"  # 替换为你的输出文件路径

# 调用函数并写入结果
csv_row_counts, total_rows = count_csv_rows_in_folder(folder_path)
write_sorted_counts_to_file(csv_row_counts, output_file)

print(f"统计结果已写入文件: {output_file}")
```
